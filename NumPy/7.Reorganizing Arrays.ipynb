Reorganizing Arrays

```python
import numpy as np
```

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)

print("curent dim: ",before.shape)
```

```python
after = before.reshape((1,8))



print(after)

print("----------------------")

print(before.reshape((8,1)))

print("----------------------")

print(before.reshape((2,4)))

print("----------------------")

print(before.reshape((4,2)))

print("----------------------")

print(before.reshape((2,2,2)))
```

```python
print(before.reshape((2,3))) # This will not work
```

```python
# Vertically stacking vectors
v1=np.array([1,2,3,4])
v2=np.array([5,6,7,8])


print(np.vstack([v1,v2,v1,v2]))
```

```python
# Horizontal  stack

h1=np.ones((2,4))
h2=np.zeros((2,2))

print(h1)
print("-------------")
print(h2)
print("-------------")

print(np.hstack((h1,h2)))
```
