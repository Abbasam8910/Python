Linear Algebra

```python
import numpy as np
```

```python
a=np.ones((2,3))
print(a)

print("----------------")


b=np.full((3,2),2)
print(b)

print("----------------")
print("Matrix Multiplication")

print(np.matmul(a,b))
```

```python
# Find the determinant

c=np.identity(3)
print(c)
print("Determinant is :")
print(np.linalg.det(c))
```
